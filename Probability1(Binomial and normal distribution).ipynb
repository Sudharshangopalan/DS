```python
# Normal distribution
# Question-3
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
a=np.random.normal(loc=0,scale=1,size=100)
plt.hist(a,bins=15)
plt.title("Normal Distribution")
plt.show()
```
