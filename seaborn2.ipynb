```python
import seaborn as sns
import seaborn.objects as so
```

```python
sns.set_theme()
```

```python
tips=sns.load_dataset("tips")
```

```python
sns.relplot(
    data=tips,
    x="total_bill", y="tip", col="time",
    hue="smoker", style="smoker", size="size",
)
```

```python
sns.relplot(data=tips,kind="scatter",hue="smoker",x="total_bill",y="tip",size="size",style="smoker")
```

```python
dots = sns.load_dataset("dots")
sns.relplot(
    data=dots, kind="line",
    x="time", y="firing_rate", col="align",
    hue="choice", size="coherence", style="choice",
    facet_kws=dict(sharex=False),
)
```

```python
fmri = sns.load_dataset("fmri")
sns.relplot(
    data=fmri, kind="line",
    x="timepoint", y="signal", col="region",
    hue="event", style="event",
)
```

```python
sns.lmplot(data=tips, x="total_bill", y="tip", col="time", hue="smoker")
```

```python
sns.displot(data=tips, x="total_bill", col="time", kde=True)
```

```python
sns.displot(data=tips, kind="ecdf",x="total_bill", col="time", rug=True)
```

```python
sns.catplot(data=tips, kind="swarm", x="day", y="total_bill", hue="smoker")
```

```python
sns.catplot(data=tips, kind="violin", x="day", y="total_bill", hue="smoker")
```

```python
sns.catplot(data=tips, kind="bar", x="day", y="total_bill", hue="smoker")
```

```python
(
    fmri
    .query("region == 'parietal' and event == 'stim'")
    .pipe(so.Plot, "timepoint", "signal")
    .add(so.Line(color=".2", linewidth=1), group="subject")
)
```

```python
p = so.Plot(fmri, "timepoint", "signal", color="region", linestyle="event")
p.add(so.Line(), so.Agg())
```

```python
(
    p
    .add(so.Line(), so.Agg())
    .add(so.Band(),so.Est(), group="event")
)
```

```python
p.add(so.Line(marker="s", edgecolor="black"), so.Agg(), linestyle=None)
```
