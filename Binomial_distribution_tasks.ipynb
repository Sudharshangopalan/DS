```python
#1.Online shopping e-commerce
# Qtn - An e-commerce company finds that 90% of orders are delivered on time.A manager randomly checks 12 orders.

a=binom.pmf(p=0.9,k=10,n=12)
print("Probability of getting exactly 10 orders:",a)

a=binom.pmf(p=0.9,k=11,n=12)
b=binom.pmf(p=0.9,k=12,n=12)
print("\nProbability of getting atleast 11 orders:",a+b)
#OR
o=binom.sf(p=0.9,n=12,k=10)
print("Probability of getting atleast 11 orders:",o)

c=binom.cdf(p=0.9,k=8,n=12)
print("\nProbability of getting fewer than 9 orders:",c)

```

```python
#2.Campus placement
# Qtn-A college reports that 65% of students get placed.A trainer selects 15 students.
#p=0.65
#n=15
#k=12

s=binom.pmf(p=0.65,n=15,k=12)
print("probabaility that exactly 12 students are placed:",s)
```

```python
s1=binom.pmf(p=0.65,n=15,k=10)
s2=binom.pmf(p=0.65,n=15,k=11)
s3=binom.pmf(p=0.65,n=15,k=12)
s4=binom.pmf(p=0.65,n=15,k=13)
s5=binom.pmf(p=0.65,n=15,k=14)
s6=binom.pmf(p=0.65,n=15,k=15)
print("probability that atleast 10 students are placed:",s1+s2+s3+s4+s5+s6)
#or
sd=binom.sf(p=0.65,n=15,k=9)
print("probability that atleast 10 students are placed:",sd)
```

```python
n=15
p=0.65
mean=n*p
print("Expected number of students placed:",mean)
```

```python
#3.Bank loan approval
# Qtn - A bank approves 70% of loan applications.An officer processes 20 applications.

l=binom.pmf(p=0.7,n=20,k=16)
print("Probability that exactly 16 loans are approved:",l)
```

```python
l1=binom.pmf(p=0.7,n=20,k=19)
l2=binom.pmf(p=0.7,n=20,k=20)
print("Probability that more than 18 loans are approved:",l1+l2)
#or
l3=binom.sf(p=0.7,n=20,k=18)
print("Probability that more than 18 loans are approved:",l3)
```

```python
print("Mean and variance")
n=20
p=0.7
mean=n*p
print("Mean:",mean)

q=1-p
variance=n*p*q
print("Variance:",variance)
```

```python
# 4.Cricket Match
# Qtn - A batsman has a 40% chance of hitting a boundary on any ball.He faces 10 balls.

ex=binom.pmf(p=0.4,n=10,k=5)
print("Probability that exactly 5 boundaries:",ex)
```

```python
at=binom.sf(p=.4,n=10,k=6)
print("Probability of atleast 7 boundaries:",at)
#or
a1=binom.pmf(p=.4,n=10,k=7)
a2=binom.pmf(p=.4,n=10,k=8)
a3=binom.pmf(p=.4,n=10,k=9)
a4=binom.pmf(p=.4,n=10,k=10)
print("Probability of atleast 7 boundaries:",a1+a2+a3+a4)
```

```python
no=binom.pmf(p=0.4,n=10,k=0)
print("Probability of no boundaries:",no)
```

```python
# 5.Manufacturing quality check.
# Qtn - A factory produces bulbs with a 5% defect rate.An inspector checks 30 bulbs.

m=binom.pmf(p=0.05,n=30,k=2)
print("Probability of exactly 2 defective bulbs:",m)
```

```python
# 6.Email marketing
# Qtn - A company knows that 25% of customers click a promotional email.The company sends emails to 18 customers.

exact=binom.pmf(p=0.25,n=18,k=6)
print("Probability of exactly 6 customers click:",exact)

at=binom.sf(p=0.25,n=18,k=3)
print("Probability that atleast 4 customers click:",at)

# expected number
n=18
p=0.25
mean=n*p
print("Expected :",mean)
```

```python
# 7.Hospital vaccination
# Qtn - A vaccine is effective in 95% of people.Doctors observe 25 patients.

exact=binom.pmf(p=0.95,n=25,k=24)
print("Probability of exactly 24 patients respond successfully:",exact)

e1=binom.pmf(p=0.95,n=25,k=25)
print("\nProbability that all 25 patients respond successfully:",e1)

atl=binom.cdf(p=0.95,n=25,k=22)
print("\nProbability of fewer than 23 patients respond successfully:",atl)
```

```python
#8.Software testing
# Qtn - A software tester finds that 92% of test cases pass successfully.A release contains 20 test cases.

s=binom.pmf(p=.92,n=20,k=18)
print("Probability that exactly 18 test cases pass:",s)

s1=binom.pmf(p=.92,n=20,k=20)
print("\nProbability that all test cases pass:",s1)

s2=binom.sf(p=.92,n=20,k=18)
print("\nProbability that atleast 19 test cases pass:",s2)
```

```python
# 9.Students attendance
# Qtn - The probability that a student attends class on a given day is 0.8.A teacher records attendance for 15 students.

std=binom.pmf(p=0.8,n=15,k=12)
print("Probability that exactly 12 students attend:",std)

std1=binom.sf(p=0.8,n=15,k=10)
print("\nProbability that more than 10 students attend:",std1)

n=15
p=0.8
std2=n*p
print("\nExpected attendance:",std2)
```

```python
# 10.Mobile App Installation
# Qtn - An advertisement convinces 30% of users to install an app.The ad is shown to 25 users.

m1=binom.pmf(p=0.3,n=25,k=8)
print("Probability that exactly 8 users install the app:",m1)

m2=binom.sf(p=0.3,n=25,k=9)
print("\nProbability that atleast 10 users install the app:",m2)

m3=binom.cdf(p=0.3,n=25,k=4)
print("\nProbability that fewer than 5 users install the app:",m3)
```
