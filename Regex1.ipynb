```python
import regex as re
```

```python
#find
text='this is my number 9876544332 '
n=re.findall(r"\d",text)
print(n)
```

```python
text='this is my email@mail.in '
n=re.findall(r'^[a-zA-Z0-9\ ]+@[a-z]+\.[a-z]+ $',text)
print(n)
```

```python
#search
text='im python developer'
py=re.search('python',text)
py
```

```python
txt="Im working at the company named google"
s=re.search('company',txt)
s
```

```python
#split
txt="Im working at the company named google"
sp=re.split('at',txt)
sp
```

```python
#sub
t="Im working at the company named google"
s=re.sub('google','amazon',t)
s
```

```python
#match
text = "Hello World"

result = re.match("Hello", text)
result
```
