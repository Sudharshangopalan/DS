```python
import numpy as np
```

```python
marks=np.random.randint(1,101,500)
marks
```

```python
highest_marks=marks[marks>90]
print("number of toppers:",len(highest_marks))
print("topper marks:",highest_marks)
```

```python
average_marks=np.mean(marks)
print("average mark:",average_marks)
```

```python
average_performer= marks[(marks >= 40) & (marks <= 75)]
print("number of average performer:",len(average_performer))
print("average performer marks:",average_performer)
```

```python
fail_mark=marks[marks<40]
print("number of failed students:",len(fail_mark))
print("failed students marks:",fail_mark)
```
