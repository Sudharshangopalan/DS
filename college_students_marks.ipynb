```python
import numpy as np
import pandas as pd
```

```python
marks=np.random.randint(0,100,500)
print(marks)
```

```python
topper=marks[marks>90]
print("num of toppers:",len(topper))
```

```python
print("topper marks:",topper)
```

```python
average_marks=np.mean(marks)
print("average_performer:",average_marks)
```

```python
average_performers=np.where((marks>=40)&(marks<=75))
print("average_performer:",len(average_performers))
```

```python
print("average_performer:",average_performers)
```

```python
failed_students=marks[marks<40]
print("num of failed students:",len(failed_students))
```

```python
print("failed student marks:",failed_students)
```
