```python
import seaborn as sns
```

```python
#SCATTER PLOT()-Relationship
tips=sns.load_dataset("tips")
tips.head(10)
```

```python
sns.scatterplot(tips,x="total_bill",y="tip")
```

```python
sns.scatterplot(tips,x="total_bill",y="tip",hue="time")
```

```python
sns.scatterplot(tips,x="total_bill",y="tip",hue="time",style="time")
```

```python
sns.scatterplot(tips,x="total_bill",y="tip",hue="day",style="time")
```

```python
sns.scatterplot(tips,x="total_bill",y="tip",hue="day")
```

```python
sns.scatterplot(tips,x="total_bill",y="tip",hue="size",palette="deep")
```

```python
tip_rate = tips.eval("tip / total_bill").rename("tip_rate")
sns.scatterplot(data=tips, x="total_bill", y="tip", hue=tip_rate)
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="size", size="size")
```

```python
#from numpy import full
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="size", size="size",
                sizes=(20,200),legend="full")
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="size", size="size",
                sizes=(20,200),hue_norm=(0,6),legend="full")
```

```python
markers = {"Lunch": "X", "Dinner": "s"}
sns.scatterplot(data=tips, x="total_bill", y="tip", style="time", markers=markers)
```

```python
#LINE PLOT()-
flights = sns.load_dataset("flights")
flights.head(10)
```

```python
jan_flights = flights.query("month == 'Jan'")
sns.lineplot(data=jan_flights, x="year", y="passengers")
```

```python
flights_wide = flights.pivot(index="year", columns="month", values="passengers")
flights_wide.head(10)
```

```python
sns.lineplot(data=flights_wide["Feb"])
```

```python
sns.lineplot(data=flights_wide)
```

```python
sns.lineplot(data=flights, x="year", y="passengers")
```

```python
sns.lineplot(data=flights, x="year", y="passengers",hue="month")
```

```python
sns.lineplot(data=flights, x="year", y="passengers",hue="month",style="month")
```

```python
sns.lineplot(data=flights, x="passengers", y="year",orient="y")
```

```python
sns.lineplot(data=flights, x="year", y="passengers",orient="x")
```

```python
fmri = sns.load_dataset("fmri")
fmri.head(10)
```

```python
#BOX PLOT()-
titanic=sns.load_dataset("titanic")
sns.boxplot(x=titanic["age"])
```

```python
titanic = sns.load_dataset("titanic")
titanic.head(10)
```

```python
sns.boxplot(data=titanic, x="age", y="class",orient="y")
```

```python
sns.boxplot(data=titanic, x="age", y="class",orient="y",hue="alive")
```

```python
sns.boxplot(data=titanic, x="class", y="age",orient="x",hue="alive",fill=0,gap=0.2)
```

```python
sns.boxplot(data=titanic, x="age", y="deck", whis=(0, 100))
```

```python
sns.boxplot(data=titanic, x="age", y="deck", width=.3)
```

```python
sns.boxplot(data=titanic, x="age", y="deck", color=".8", linecolor="red", linewidth=.75)
```

```python
ax = sns.boxplot(x=titanic["age"].round(-1), y=titanic["fare"], native_scale=1)
ax.axvline(35, color=".3", dashes=(3, 2))
```

```python
sns.boxplot(
    data=titanic, x="age", y="class",
    notch=True, showcaps=0,
    flierprops={"marker": "x"},
    boxprops={"facecolor": (.3, .5, .7, .5)},
    medianprops={"color": "r", "linewidth": 2},
)
```

```python
df = sns.load_dataset("titanic")
sns.violinplot(x=df["age"])
```

```python
sns.violinplot(data=df, x="class", y="age")
```

```python
sns.violinplot(data=df, x="class", y="age", hue="alive", fill=False)
```

```python
sns.violinplot(data=df, x="class", y="age", hue="alive", split=True, gap=.3, inner="quart")
```

```python
sns.violinplot(data=df, x="age", y="deck", inner="point", density_norm="count")
```

```python
sns.violinplot(data=df, x="age", y="alive", cut=0, inner="stick")
```

```python
sns.violinplot(x=df["age"].round(-1) + 5, y=df["fare"], native_scale=True)
```
