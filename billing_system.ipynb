```python
import numpy as np
```

```python
products = np.array(["Rice", "Milk", "Eggs", "Sugar"])
```

```python
quantity = np.array([15, 20, 10, 5])
```

```python
price = np.array([55, 25, 6, 45])
```

```python
daily_revenue = np.array([2500, 3200, 2800, 3500, 4000, 3700, 4200])
```

```python
sales = quantity * price
total_sales = np.sum(sales)

print("Total Sales =", total_sales)
```

```python
maxsale = np.argmax(quantity)

print("Highest-Selling Product:")
print("Product Name =", products[maxsale])
print("Quantity Sold =", quantity[maxsale])
```

```python
average_revenue = np.mean(sales)
print("daily revenue:",average_revenue)
```
