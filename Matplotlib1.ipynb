```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#line graph
plt.plot([2.3,6,1,12,8,10],marker='d',ms=5,mec='black',mfc='green')
plt.show()
```

```python
#bar chart
product=['jewels','cloths','milk','icecream']
quantity=[1200,6500,4500,5221]
plt.bar(product,quantity,color='red',edgecolor='black')
plt.title("Bar graph for product vs quantity")
plt.xlabel('product')
plt.ylabel('quantity')
plt.show()
```

```python
#pie chart
shares=[35,45,55,30]
product=['jewels','cloths','milk','icecream']
plt.pie(shares,labels=product,colors=['blue','red','pink','green'],autopct='%1.2f%%',startangle=90)
plt.title("Bar graph for product vs shares")
plt.show()
```

```python
#histogram
marks=[35,45,50,60,70,80,95,99]
plt.hist(marks,color="red",bins=15,edgecolor="blue")
plt.title("Histogram for marks")
plt.xlabel("marks")
plt.ylabel("counts")
plt.show()
```

```python
#Box plot
chennai=[35,40,50,65,35,25,30]
salem=[34,25,68,67,45,34,57,]
covai=[15,23,24,3,56,67,50]
plt.title("Box plot ")
plt.boxplot([chennai,salem,covai],labels=["chennai","salem","covai"])
plt.show()
```

```python
#scatter plot
age=[18,20,25,35,50,76]
Fare=[1200,140,500,390,3200,250]
plt.scatter(age,Fare,marker='s')
plt.show()
```

```python
#area plot
s=[2.3,6,1,12,8,10]
d=[1,2,3,4,5,6]
plt.plot(s,d,marker='d',ms=5,mec='black',mfc='green')
plt.fill_between(s,d)
plt.show()
```

```python
#sub plot

plt.subplot(2,2,1)
plt.scatter(age,Fare,marker='s')
plt.subplot(2,2,2)
plt.bar(product,quantity,color='red',edgecolor='black')
plt.subplot(2,2,3)
plt.boxplot([chennai,salem,covai],labels=["chennai","salem","covai"])
plt.subplot(2,2,4)
plt.pie(shares,labels=product,colors=['blue','red','pink','green'],autopct='%1.2f%%',startangle=90)
plt.figure(figsize=(10,5))
plt.show()
```
