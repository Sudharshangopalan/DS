```python
print("count flowers in each species")
i['species'].value_counts()
```

```python
result =i.groupby('species')['petal_length'].mean()
print("average petal length for each species  ")
result
```

```python
max_sepal_length = i['sepal_length'].max()
print("max_sepal_length : ",max_sepal_length)
```

```python
max_flower = i[i['sepal_length'] == i['sepal_length'].max()]
max_flower
```

```python
sort = i.sort_values(by='petal_width',ascending=False)
print("Sort by petal width descending  ")
sort
```

```python
filter=i[i["petal_length"]>5]
print("Filter flowers with petal_length greater than 5 are ")
filter
```

```python
print("Adding new column as total_size")
i["total_size"]=i["sepal_length"]+i["petal_length"]
i
```

```python
import numpy as np
```

```python
petal_array = i['petal_length'].to_numpy()
print("petal array ")
petal_array
```

```python
print("Mean:", np.mean(petal_array))
```

```python
print("Median:", np.median(petal_array))
```

```python
print("Standard Deviation:", np.std(petal_array))
```

```python
print("maximum value of petal array")
petal_array.max()
```

```python
print("minimum value of petal array")
petal_array.min()
```

```python
print("Max:", np.max(petal_array))
print("Min:", np.min(petal_array))
```

```python
normalized = (petal_array - np.min(petal_array)) / (np.max(petal_array) - np.min(petal_array))

print(normalized)
```
